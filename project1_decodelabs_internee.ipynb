```python
#project 1 chat bot

responses={
    "hello":"hello! how are you",
    "hi":"hey there",
    "how are you": "I am fine. Thank you!",
    "bye": "Goodbye!",
    "exit": "Goodbye!",

}


print("Chat bot here ! ")
print("Type 'exit' to quit ")


while True:
  inp=input("Go ahead! ").lower().strip()
  reply=responses.get(inp)
  if reply is None:
    reply="I didnt get it !"
  print("Bot reply: ",reply)
  if inp=="exit":
    break




```
